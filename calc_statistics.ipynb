```python
import pandas as pd
import numpy as np

# Read dataset
df = pd.read_csv("students.csv")

# Total Marks
df["Total"] = df["Math"] + df["Physics"] + df["Chemistry"]

# Average Marks
df["Average"] = df["Total"] / 3

# Percentage
df["Percentage"] = (df["Total"] / 300) * 100

# Subject Averages
math_avg = np.mean(df["Math"])
physics_avg = np.mean(df["Physics"])
chemistry_avg = np.mean(df["Chemistry"])

print("Math Average:", math_avg)
print("Physics Average:", physics_avg)
print("Chemistry Average:", chemistry_avg)

# Highest Average Subject
subject_averages = {
    "Math": math_avg,
    "Physics": physics_avg,
    "Chemistry": chemistry_avg
}

highest_subject = max(subject_averages, key=subject_averages.get)
print("\nHighest Average Subject:", highest_subject)

# Lowest Average Subject
lowest_subject = min(subject_averages, key=subject_averages.get)
print("Lowest Average Subject:", lowest_subject)

# Standard Deviation
print("\nStandard Deviation")
print("Math Std Dev:", np.std(df["Math"]))
print("Physics Std Dev:", np.std(df["Physics"]))
print("Chemistry Std Dev:", np.std(df["Chemistry"]))

# Basic Statistics
print("\nBasic Statistics")
print(df.describe())

# Save final file
df.to_csv("student_statistics.csv", index=False)

print("\nStatistics saved successfully")
```
