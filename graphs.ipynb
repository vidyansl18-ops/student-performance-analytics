```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("students.csv.txt")

df.head()
```

```python
# Calculate subject averages

averages = [
    df["Math"].mean(),
    df["Physics"].mean(),
    df["Chemistry"].mean()
]

subjects = ["Math", "Physics", "Chemistry"]

# Create bar graph

sns.barplot(x=subjects, y=averages)
plt.ylim(60,75)

plt.title("Average Subject Scores")
plt.savefig("average_scores.png")
# Show graph

plt.show()
```

```python
# Student names
names = df["Name"]

# Attendance values
attendance = df["Attendance"]

# Create graph
sns.barplot(x=names, y=attendance)

# Title
plt.title("Student Attendance")

# Y-axis range
plt.ylim(60, 100)
plt.xticks(rotation=45)
plt.savefig("student_attendance.png")
# Show graph
plt.show()
```

```python
# Create total marks column

df["Total"] = (
    df["Math"] +
    df["Physics"] +
    df["Chemistry"]
)

# Create graph

sns.barplot(x=df["Name"], y=df["Total"])

# Title

plt.title("Total Marks of Students")

# Rotate names

plt.xticks(rotation=45)
plt.savefig("total_marks.png")
# Show graph

plt.show()
```

```python
# Set figure size

plt.figure(figsize=(8,5))

# Plot lines

plt.plot(df["Name"], df["Math"], marker='o', label="Math")
plt.plot(df["Name"], df["Physics"], marker='o', label="Physics")
plt.plot(df["Name"], df["Chemistry"], marker='o', label="Chemistry")

# Title

plt.title("Subject-wise Student Performance")

# Labels

plt.xlabel("Students")
plt.ylabel("Marks")

# Rotate names

plt.xticks(rotation=45)

# Show legend

plt.legend()

# Show graph
plt.savefig("s_W_performance.png")
plt.show()
```

```python
# Create total column

df["Total"] = (
    df["Math"] +
    df["Physics"] +
    df["Chemistry"]
)

# Create scatter plot

sns.scatterplot(
    x=df["Attendance"],
    y=df["Total"]
)

# Title

plt.title("Attendance vs Total Marks")

# Labels

plt.xlabel("Attendance")
plt.ylabel("Total Marks")

# Show graph
plt.savefig("a_vs_tm.png")
plt.show()
```
